```python
###### Generate_FAG_Relative_Parents_Table.py
# 6/26/2025


from bs4 import BeautifulSoup 
import datetime
from datetime import datetime as dt
from datetime import date 
from datetime import timedelta
import time
import re
from pandas import read_csv
import pandas as pd
# from selenium import webdriver
from seleniumwire import webdriver
from selenium.webdriver.common.by import By
import json
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
import sys
import random

def display_dataframe_segments(dataframe_df):
    for i in range(0,len(dataframe_df),50):
        print(dataframe_df.iloc[i:i+50])
        
def clean_dataframe(dataframe_df,column):
    delete_list=['Pvt ','PTE ','Dr. ','Rev ','MAJ ','GEN ','Mrs ','CPT ',',PFC ','SGT ','Col ','LT ','Dr ',\
                ' Sr',' Jr','\.','Minor ','Infant ','Boy ','Girl ','Twins ']
    for item in delete_list:
        dataframe_df[column]= dataframe_df[column].astype(str).replace(item, ' ', regex=True).str.strip()
    return dataframe_df        

print("***Generate_FAG_Relative_index--- Started *****")
DirMain="C:\\Users\\jrozm\\OneDrive\\Benjamin\\"

grave_source='ABMC'
# grave_source='Arlington'

# relative_list=['Parents','Siblings','Spouses','Children']
relative_list=['Parents']
fileInput=grave_source+"_FAG_Mereged"

fileIn=DirMain+"Analysis\\"+fileInput+".csv"

relative_in_df=pd.read_csv(fileIn, encoding='ISO-8859-1')

relative_in_df['Birth_index']=relative_in_df['Birth_index'].fillna(100001010).astype(int)
relative_in_df['Death_index']=relative_in_df['Death_index'].fillna(100001010).astype(int)
relative_in_df['birthDate'] = pd.to_datetime(relative_in_df['birthDate'].fillna('01/01/1000'),format='%m/%d/%Y',errors='coerce')
relative_in_df['deathDate'] = pd.to_datetime(relative_in_df['deathDate'].fillna('01/01/1000'),format='%m/%d/%Y',errors='coerce')


field_list=['Name_Index', 'Last_Name', 'First_Name', 'Middle_Name', 'birthDate', 'Birth_index', 'deathDate',\
            'Death_index', 'birthCityName', 'birthStateName', 'birthCountryName', 'links', 'Parents', 'Spouses', 'Siblings', 'Children']
field_list_set=set(field_list)

for relative_type in relative_list:
    relative_index=relative_type+'_name_index'
    fileOutput=grave_source+"_Relative_"+relative_type+"_Table_Out"
    fileOut=DirMain+"Analysis\\"+fileOutput+".csv"

    fileOutput=grave_source+"_Relative_"+relative_type+"_Table_Out"
    fileOut=DirMain+"Analysis\\"+fileOutput+".csv"
    
    relative_in_df[relative_type]=relative_in_df[relative_type].str.strip()
    relative_df=relative_in_df.copy()

    relative_df = relative_df.dropna(subset=['Parents'])
    relative_df.reset_index(drop=True, inplace = True)

    field_list_relative=['Name_Index','Last_Name','First_Name','Middle_Name','birthDate','Birth_index','deathDate','Death_index',\
             'birthCityName','birthStateName','parent_1','parent_2']      

    field_list_split=field_list.copy()
    field_list_split.append('split')

    relative_df=clean_dataframe(relative_df,relative_type)
    relative_df_split=relative_df[relative_type].str.split("*")
 
    relative_df['split']=relative_df_split
    relative_df['parent_1']=relative_df['split'].str[0].str.strip()
    relative_df['parent_2']=relative_df['split'].str[1].str.strip()

    relative_parent_df_split1=relative_df['parent_1'].str.split(" ")
    relative_df['parent_1']=relative_parent_df_split1.str[-1]+'-'+relative_parent_df_split1.str[0]      
    relative_parent_df_split2=relative_df['parent_2'].str.split(" ")
    relative_df['parent_2']=relative_parent_df_split2.str[-1]+'-'+relative_parent_df_split2.str[0]        
   
    relative_df=relative_df[field_list_relative]
    relative_df.to_csv(fileOut,mode="w", index=False,header=True)
    print("\n")
    print('Uploaded To: ',fileOut)
print("***Generate_FAG_Relative_index--- Complete *****")
```
