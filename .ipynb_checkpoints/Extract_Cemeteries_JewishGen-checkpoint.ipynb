```python
# Extract_Cemeteries_JewishGen.py


from bs4 import BeautifulSoup 
import datetime
from datetime import datetime as dt
from datetime import date 
from datetime import timedelta
import time
import re
from pandas import read_csv
import pandas as pd
import json
import sys
import math
import numpy as np
import urllib3
import inspect

def maximize_screen():
    from IPython.display import display, HTML
    display(HTML("<style>:root { --jp-notebook-max-width: 100% !important; }</style>"))

def timestamp():
    today=str(datetime.date.today()).split('-')
    year=today[0]
    month=today[1]
    day=today[2]
    a=str(datetime.datetime.now())
    time_split=re.split(':| ',a)
    hour=time_split[1]
    timestamp=year+month+day+hour
    return timestamp

def clean_dataframe(graves_df):
    global field_list
    for column in field_list:
        graves_df[column]= graves_df[column].astype(str).replace(r"\,", ' * ', regex=True).str.strip()
        graves_df[column]= graves_df[column].astype(str).replace(r"\'", '', regex=True).str.strip()
    delete_list=[r'\"',r"\'",r'\?',r'\[',r'\]']
    for item in delete_list:
        for column in field_list:
            graves_df[column]= graves_df[column].astype(str).replace(item, ' ', regex=True).str.strip()
    return graves_df


blank_line=['                                                                                                                                                                      ']

maximize_screen()
timestamp=timestamp()
collection_date0=datetime.datetime.now().strftime("%m/%d/%Y")
collection_date_dt=pd.to_datetime(collection_date0)
collection_date=collection_date0

start_time=datetime.datetime.now()
event_time=start_time

process_name='Extract Cemeteries JewishGen'
print('**Starting *** ',process_name,' *** , Time:',start_time,' **\n')
DirMain="C:\\Users\\jrozm\\OneDrive\\Benjamin\\"



fileInput="Jewish_Gen_Jewish_Cemeteries"
fileOutput="Jewish_Gen_Jewish_Cemeteries"

# if append==1:fileIn='JewishCemeteries_CountyCorrelated_X'


fileIn=DirMain+"DataFiles\\"+fileInput+".html"
fileOut=DirMain+"OutputData\\Cemeteries\\"+fileOutput+"_"+timestamp+".csv"

fileOut0=DirMain+"OutputData\\Cemeteries\\"+fileOutput+".csv"
fileOutInt=DirMain+"OutputData\\Cemeteries\\"+fileOutput+"_Interim.csv"

field_list=['country','state','city','cemetery_name','cemetery_id','graves']

# Let's assume you have the HTML in a string called html_text
with open(fileIn, "r", encoding="utf-8") as f:Cemeteries_html = f.read()

beautifulSoupText = BeautifulSoup(Cemeteries_html, "html.parser")
cemetery_df = pd.DataFrame([])

# Traverse down the tree
cemetery_links = beautifulSoupText.find_all('a', href=re.compile(r'/databases/cemetery/jowbrshow\.php'))

cemetery_count=0
for link in cemetery_links:
    # Extract the parent list item which contains the link
    li = link.parent
    # print('\nlink:',link)
    # print('li:',li)
    # Extract the cemetery name from the link text
    cemetery_name = link.text.strip()
    # Extract the ID from the link href
    cemetery_id = link['href'].split('ID=')[1]         
    
    # Extract stats using regex from the rest of the text in the list item
    full_text = li.get_text()
    stats_text = full_text.replace(cemetery_name, '').strip()       
    # Extract burials, photos, and date from the text
    burials_match = re.search(r'(\d+) burials', stats_text)
    photos_match = re.search(r'(\d+) photos', stats_text)
    date_match = re.search(r'data online (\d+ \w+ \d+)', stats_text)
    
    burials = int(burials_match.group(1)) if burials_match else 0
    photos = int(photos_match.group(1)) if photos_match else 0
    date_online = date_match.group(1) if date_match else None
     
    # Find the city by traversing up the DOM
    # This assumes the city is in an "LI" with id="fh" above each cemetery
    # print("li.parent.find_previous('li', id='fh'):",li.parent.find_previous('li', id='fh'))
    # input('')
    city_element = li.parent.find_previous('li', id='fh')
    try:
        city = city_element.get_text().split('City:')[1].split('(')[0].strip() if city_element else None
    except:
        continue
        
    # Find the region by going up another level
    region_element = city_element.parent.find_previous('li', id='fh') if city_element else None
    region = region_element.get_text().split('Region:')[1].split('(')[0].strip() if region_element else None

    cemetery_line_df=pd.DataFrame([country,region,city,cemetery_name,cemetery_id,burials]).T
    cemetery_line_df.columns=field_list
    cemetery_line_df=clean_dataframe(cemetery_line_df)

    cemetery_df = cemetery_df._append(cemetery_line_df, ignore_index = True) 
    if cemetery_count%200==0:print(f"    --- {process_name:<20}• State: {region} • City: {city} • cemetery count: {cemetery_count} ")
    cemetery_count+=1
print(f" **{process_name:<20}• US  • cemetery count: {us_cemetery_count} ")
cemetery_df.to_csv(fileOut,mode="w", index=False,header=True)
cemetery_df.to_csv(fileOut0,mode="w", index=False,header=True) 
print("\n")
print('Uploaded To: ',fileOut0)
```
