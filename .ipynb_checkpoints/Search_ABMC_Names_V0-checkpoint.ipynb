```python
# Search_ABMC_Names.py
# 5/20/2025

from bs4 import BeautifulSoup 
import datetime
from datetime import datetime as dt
from datetime import date 
from datetime import timedelta
import time
import re
from pandas import read_csv
import pandas as pd
# from selenium import webdriver
from seleniumwire import webdriver
from selenium.webdriver.common.by import By
import json
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
import sys
import random
from zenrows import ZenRowsClient
import requests
import math
import numpy as np
import urllib3
import inspect
from geopy.distance import distance
# from IPython.display import display


# def maximize_screen():
#     from IPython.display import display, HTML
#     display(HTML("<style>:root { --jp-notebook-max-width: 100% !important; }</style>"))

def left(s, amount):
    return s[:amount]
def right(s, amount):
    return s[-amount:]
def mid(s, offset, amount):
    return s[offset:offset+amount]

def maximize_screen():
    from IPython.display import display, HTML
    display(HTML("<style>:root { --jp-notebook-max-width: 100% !important; }</style>"))

def timestamp():
    today=str(datetime.date.today()).split('-')
    year=today[0]
    month=today[1]
    day=today[2]
    a=str(datetime.datetime.now())
    time_split=re.split(':| ',a)
    hour=time_split[1]
    timestamp=year+month+day+hour
    return timestamp

def convert_date_z(date): 
    date_t= pd.to_datetime(date, utc=True)
    date_out=date_t.dt.strftime("%m/%d/%Y")
    return date_out

def convert_datetime_t(event_date_t):
    no_time='2000-01-01T00:00:00'
    no_time_dt=dt.strptime(no_time, "%Y-%m-%dT%H:%M:%S") 
    event_date_t = event_date_t.fillna(no_time)
    event_date_dt=pd.to_datetime(event_date_t, format="%Y-%m-%dT%H:%M:%S")
    event_date_index=event_date_dt.dt.strftime("%Y%m%d")
    event_date =event_date_dt.dt.strftime("%m/%d/%Y")
    event_date_rnd_dt = event_date_dt + timedelta(minutes=30)
    event_time_index=event_date_rnd_dt.dt.strftime('%H')  
    event_time = event_date_dt.dt.strftime('%H:%M')
    return     event_date,event_date_index,event_time,event_time_index 



def random_sleep(call_count): 
    global sleep_time_pause,sleep_range
    if call_count%100==0:
        sleep_time =sleep_time_pause
        now_date = datetime.datetime.now() + timedelta(seconds=sleep_time)
        sleep_end_date=now_date.strftime("%m/%d/%Y %H:%M:%S")    
        print(f"\t\t\tSleeping for {sleep_time:.0f} seconds • Sleeping till: {sleep_end_date}")
    else:
        sleep_time = random.uniform(sleep_range[0],sleep_range[1])
    return sleep_time 

def get_total_cemeteries(url):
    global my_header
    response = requests.request("GET", url, headers=my_header) 
    status_code=response.status_code
    url_source = response.text 
    
    beautifulSoupText = BeautifulSoup(url_source, 'html.parser') 
    specific_tag_text = beautifulSoupText.get_text()
    capture_json=json.loads(specific_tag_text) 
    total_cemeteries=capture_json['data']['cemeteries']['total']
    return total_cemeteries

def get_grave_count(url):
    response = requests.request("GET", url, headers=my_header) 
    status_code=response.status_code
    url_source = response.text 
    beautifulSoupText = BeautifulSoup(url_source, 'html.parser') 
    specific_tag_text = beautifulSoupText.get_text()
    capture_json=json.loads(specific_tag_text) 
    grave_count=capture_json['total']
    return capture_json,grave_count
    
def extract_grave_data(capture_json):
    global field_list_set,field_list
    collection_df=pd.json_normalize(capture_json['collection'])
    collection_columns_set=set(list(collection_df))
    missing_columns=field_list_set-collection_columns_set
    collection_df[list(missing_columns)]=np.nan
    grave_frame_df=collection_df[field_list]
    return grave_frame_df

def clean_dataframe(graves_df):
    global field_list
    for column in field_list:
        graves_df[column]= graves_df[column].astype(str).replace(r"\,", ' * ', regex=True).str.strip()
        graves_df[column]= graves_df[column].astype(str).replace(r"\'", '', regex=True).str.strip()
    delete_list=[r'\"',r"\'",r'\?',r'\[',r'\]']
    for item in delete_list:
        for column in field_list:
            graves_df[column]= graves_df[column].astype(str).replace(item, ' ', regex=True).str.strip()
    return graves_df

def get_cluster_cemetery_list(name,name_lat,name_lon):
    global jewish_radius,non_jewish_radius,matched_cemeteries_df_header
    name_location = [name_lat,name_lon] 
    jewish_cemetery_count=0
    jewish_grave_count=0
    matched_cemeteries_df=pd.DataFrame([])
    for jewish_index, jewish_row in jewish_cemetery_df.iterrows():
        jewish_distance= int(distance(name_location, (jewish_row['cemetery_lat'], jewish_row['cemetery_lon'])).miles)
        if jewish_distance>=jewish_radius: continue
        jewish_cemetery_count+=1
        matched_cemeteries_line_df=pd.DataFrame(['Jewish',jewish_row['cemetery_id'],jewish_row['cemetery_name'],jewish_distance, jewish_row['city'],\
                                                 jewish_row['state'],jewish_row['cemetery_lat'],jewish_row['cemetery_lon'],jewish_row['grave_count']]).T
        jewish_grave_count+=jewish_row['grave_count']                                                                                                              
        matched_cemeteries_line_df.columns=matched_cemeteries_df_header
        matched_cemeteries_df = matched_cemeteries_df._append(matched_cemeteries_line_df, ignore_index = True) 
    print(f"---For Name = {name} •  Number of Jewish Cemeteries Within {jewish_radius} miles: {jewish_cemetery_count} • Total Jewish Grave Count: {jewish_grave_count}")
    
    non_jewish_cemetery_count=0
    non_jewish_grave_count=0
    total_cemetery_count=len(non_jewish_cemetery_df)
    for non_jewish_index, non_jewish_row in non_jewish_cemetery_df.iterrows():
        non_jewish_distance= int(distance(name_location, (non_jewish_row['cemetery_lat'], non_jewish_row['cemetery_lon'])).miles)
        if non_jewish_index%20000==0:print(f"  ..... Checking Cemeteries: {non_jewish_index}/{total_cemetery_count}")
        if non_jewish_distance>=non_jewish_radius: continue
        non_jewish_cemetery_count+=1
        matched_cemeteries_line_df=pd.DataFrame(['Non_Jewish',non_jewish_row['cemetery_id'],non_jewish_row['cemetery_name'],non_jewish_distance, non_jewish_row['city'],\
                                                 non_jewish_row['state'],non_jewish_row['cemetery_lat'],non_jewish_row['cemetery_lon'],non_jewish_row['grave_count']]).T
        non_jewish_grave_count+=non_jewish_row['grave_count'] 
        matched_cemeteries_line_df.columns=matched_cemeteries_df_header
        matched_cemeteries_df = matched_cemeteries_df._append(matched_cemeteries_line_df, ignore_index = True) 
    print(f"---For Name = {name} •  Number of Non-Jewish Cemeteries Within {non_jewish_radius} miles: {non_jewish_cemetery_count} • Total Non-Jewish Grave Count: {non_jewish_grave_count}\n")
    print(f"**For Name = {name} • Number of Jewish/Non-Jewish Cemeteries Within {jewish_radius}/{non_jewish_radius} miles: {jewish_cemetery_count}/{non_jewish_cemetery_count} • Grave Counts: {jewish_grave_count}/{non_jewish_grave_count}")
    return matched_cemeteries_df
    



#*************** Append  *******************************
append=0
# append_flag=input('To Append Press 1')
# if append_flag=='1':append=1
#*************** Append  *******************************

maximize_screen()
timestamp=timestamp()
collection_date0=datetime.datetime.now().strftime("%m/%d/%Y")
collection_date_dt=pd.to_datetime(collection_date0)
collection_date=collection_date0

start_time=datetime.datetime.now()
event_time=start_time

process_name='Extract Jewish ABMC Graves '
print('**Starting *** ',process_name,' *** , Time:',start_time,' **\n')
DirMain="C:\\Users\\jrozm\\OneDrive\\Benjamin\\"



fileIn="Names_List"
fileJewishIn="Jewish_Cluster_Cemetery_List"
fileNonJewishIn="NonJewish_Cluster_Cemetery_List"

fileOutput="Benjamin_Graves_Jewish_Names_Test"

# if append==1:fileIn='Benjamin_Cemeteries_Jewish_Test_X'
fileInJ=DirMain+"Data\\"+fileJewishIn+".csv"
fileInN=DirMain+"Data\\"+fileNonJewishIn+".csv"
fileNamesIn=DirMain+"Analysis\\"+fileIn+".csv"

fileOut=DirMain+"Analysis\\SearchResults\\"+fileOutput+"_"+timestamp+".csv"
fileOut0=DirMain+"Analysis\\SearchResults\\"+fileOutput+".csv"
fileOutInt=DirMain+"Analysis\\SearchResults\\"+fileOutput+"_Interim.csv"

jewish_cemetery_df=pd.read_csv(fileInJ)
non_jewish_cemetery_df=pd.read_csv(fileInN)

cemetery_field_list=['cemetery_id','grave_count','cemetery_name','city','county	state','stateAbbreviation','county_id','state_id','cemetery_lat','cemetery_lon']


names_df=pd.read_csv(fileNamesIn)
name_list=names_df["Name"]
name_lat_list=names_df["lat"]
name_lon_list=names_df["lon"]

matched_cemeteries_df_header=['type','cemetery_id','cemetery_name','distance','city','state','lat','lon','grave_count']

my_header = {
  'priority': 'u=1, i',
  'sec-ch-ua': '"Google Chrome";v="135", "Not-A.Brand";v="8", "Chromium";v="135"',
  'sec-ch-ua-platform': '"Windows"',
  'sec-fetch-mode': 'cors',
  'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36',
  'Cookie': '__cf_bm=O55WB_mmMt72i8KoQuzWliizNTb6J7zg2RVHXqbn.vc-1745620049-1.0.1.1-IDSlPNW4kRjBMAr5n8M2obOoGp9GAod_BHbB997Ze2F0YUm2DvAFh5TaijsQhPWR0iyIyFoaSy.p1YDke07ZttHHC83SJpQCEBFmaeImKIc; _cfuvid=zhvP.IczaAr9sd4YVjWjTqBayGxsei_Ntl0ANgSCCBg-1745620049033-0.0.1.1-604800000; ipLoc=us; preferredLanguage=en'
}

field_list=['lastName','firstName','fullName','birthYear','deathYear','birthMonth','deathMonth','birthDate','deathDate',\
    'cemeteryName','cemeteryId','cityName','stateName','maidenName','Parents','Children', 'Siblings','Spouses']
field_list_set=set(field_list)
if append==1:print('**** Appending ****')    
if append!=1:graves_df=pd.DataFrame([])  


url0 = "https://www.findagrave.com/cemetery/"
url1 ="/memorial-search?ajax=true&skip="
url2="&limit=100&firstname=&middlename=&lastname="
url3="&birthyear=&birthyearfilter=&deathyear=1950&deathyearfilter=before"

date_df=pd.DataFrame([])
sleep_range=[3,8]
sleep_time_pause=1*60

jewish_radius=20
non_jewish_radius=10
get_counter=0
name_count=0
for name in name_list:
    name_lat=name_lat_list.iloc[name_count]
    name_lon=name_lon_list.iloc[name_count]
    matched_cemeteries_df=get_cluster_cemetery_list(name,name_lat,name_lon)
    cemetery_search_count=0
    for cemetery_id in matched_cemeteries_df['cemetery_id']:
        type=matched_cemeteries_df['type'].iloc[cemetery_search_count]
        cemetery_search_count+=1
        frame_count=0
        name_count+=1
        grave_count=100000
        grave_frame_count=0
        while grave_frame_count<grave_count:
            frame_count+=1
            url=url0+str(cemetery_id)+url1+str(grave_frame_count)+url2+name+url3

            capture_json,grave_count=get_grave_count(url)
            grave_frame_df=extract_grave_data(capture_json)
            grave_frame_df['type']=type

            
            grave_frame_df=clean_dataframe(grave_frame_df)
            graves_df = graves_df._append(grave_frame_df, ignore_index = True) 
            grave_frame_count+=len(grave_frame_df)
            
            graves_df=clean_dataframe(graves_df)
            graves_df.to_csv(fileOutInt,mode="w", index=False,header=True)
            print(grave_frame_df)
            input('')
            print(f"    --- {process_name:<20}• Name: {name}  • grave count: {grave_frame_count}/{grave_count} ")
            get_counter+=1
            sleep_time=random_sleep(get_counter)       
            time.sleep(sleep_time) 
            
        # graves_df.to_csv(fileOutInt,mode="w", index=False,header=True)
        end_time=datetime.datetime.now()
        run_processing_time=end_time-event_time
        event_time=end_time
        run_processing_time_seconds=int(run_processing_time.total_seconds())
        processing_time=end_time-start_time
        processing_time_minutes=int(processing_time.total_seconds())/60  
        
        print(f"***{process_name:<20}• Name: {name}  • grave count: {grave_frame_count} ")
        
end_time=datetime.datetime.now()
processing_time=end_time-start_time
processing_time_seconds=int(processing_time.total_seconds())
print(f"**** {process_name}   Processing Time: {processing_time_seconds} Sec")   

graves_df.to_csv(fileOut,mode="w", index=False,header=True)
graves_df.to_csv(fileOut0,mode="w", index=False,header=True) 
print("\n")
print('Uploaded To: ',fileOut0)
```
