```python
# Plot_Jewish_Cemeteries.py

import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import datetime
from datetime import datetime as dt
from datetime import date 
from datetime import timedelta
import time
import re
from pandas import read_csv
import pandas as pd
import json
import random
import math
import numpy as np
import urllib3
import inspect
import folium
import pandas as pd
from IPython.display import display
from geopy.distance import distance

def maximize_screen():
    from IPython.display import display, HTML
    display(HTML("<style>:root { --jp-notebook-max-width: 100% !important; }</style>"))

def timestamp():
    today=str(datetime.date.today()).split('-')
    year=today[0]
    month=today[1]
    day=today[2]
    a=str(datetime.datetime.now())
    time_split=re.split(':| ',a)
    hour=time_split[1]
    timestamp=year+month+day+hour
    return timestamp

maximize_screen()
timestamp=timestamp()
collection_date0=datetime.datetime.now().strftime("%m/%d/%Y")
collection_date_dt=pd.to_datetime(collection_date0)
collection_date=collection_date0

start_time=datetime.datetime.now()
event_time=start_time

# print('**Starting *** ',process_name,' *** , Time:',start_time,' **\n')
DirMain="C:\\Users\\jrozm\\OneDrive\\Benjamin\\"

fileInput="JewishCemeteryCoordinates_Plus"
fileInputCircles="Jewish_Cemetery_Cities_Circles"
fileOutput="JewishCemetery_Site_Map"

# if append==1:fileIn='Jewish_Counties_Coordinates_X'


fileJewishIn=DirMain+"OutputData\\Cemeteries\\"+fileInput+".csv"
fileCirclesIn=DirMain+"Analysis\\Maps\\"+fileInputCircles+".csv"
fileMapOut=DirMain+"Analysis\\Maps\\"+fileOutput+"_"+timestamp+".html"

Jewish_Cemetery_Coordinates=pd.read_csv(fileJewishIn)
Jewish_Cemetery_Circles=pd.read_csv(fileCirclesIn)



cemetery_id_list=Jewish_Cemetery_Coordinates["cemetery_id"]
cemetery_name_list=Jewish_Cemetery_Coordinates["cemetery_name"]
county_id_list=Jewish_Cemetery_Coordinates["county_id"]
county_list=Jewish_Cemetery_Coordinates["county"]

grave_count_list=Jewish_Cemetery_Coordinates["grave_count"]
cemetery_city_list=Jewish_Cemetery_Coordinates["city"]

cemetery_lat_list=Jewish_Cemetery_Coordinates["cemetery_lat"]
cemetery_lon_list=Jewish_Cemetery_Coordinates["cemetery_lon"]
coordinates_list=Jewish_Cemetery_Coordinates[['cemetery_lat','cemetery_lon']]


folium_colors=['red', 'blue', 'green', 'orange', 'purple', 'darkred', 'lightred',\
'beige', 'darkblue', 'darkgreen', 'cadetblue', 'darkpurple',\
'white', 'pink', 'lightblue', 'lightgreen', 'gray', 'black','lightgray']

# Initialize map centered in the U.S.
m = folium.Map(location=[38.0, -85.0], zoom_start=5)
miles2meters=1609.34
county_cemetery_counts_thresh=2
# Add markers
for _, row in Jewish_Cemetery_Coordinates.iterrows():
    info = f"{row['cemetery_name']} - {row['cemetery_id']} ({row['city']}, {row['state']}, Grave Counts: {row['grave_count']}, {row['county_id']}, {row['state_id']})"
    county_cemetery_counts=row['County_Cemetery_Counts']
    dot_color='red'
    if county_cemetery_counts<=county_cemetery_counts_thresh:dot_color='blue'
    folium.CircleMarker(
        location=[row['cemetery_lat'], row['cemetery_lon']],
        radius=.01,                     # dot size
        color=dot_color,                  # border color
        fill=True,
        fill_color='red',             # fill color
        fill_opacity=0.8,
        tooltip=info,  # appears on hover
        # popup=folium.Popup(info, max_width=300)  # appears on click
    ).add_to(m)



for _, circle_row in Jewish_Cemetery_Circles.iterrows():
    cluster_info = f"{circle_row['city']} - {circle_row['Radius']} miles"
    folium.Circle(
        location=[circle_row['City_Lat'], circle_row['City_Lon']],
        radius=circle_row['Radius']*miles2meters,  # 100 miles in meters
        color='black',
        fill=False,
        weight=1,        # thickness of the outline in pixels
        fill_opacity=0.2,
        tooltip=cluster_info,  # appears on hover
        ).add_to(m)
title_html = '''
     <h3 align="center" style="font-size:16px"><b>Jewish Cemetery Locations</b></h3>
     '''
m.get_root().html.add_child(folium.Element(title_html))
m.add_child(folium.LatLngPopup())

m.save(fileMapOut)
display(m)

```
