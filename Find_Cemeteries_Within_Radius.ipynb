```python
# Find_Cemeteries_Within_Radius
from geopy.distance import distance
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import datetime
from datetime import datetime as dt
from datetime import date 
from datetime import timedelta
import time
import re
from pandas import read_csv
import pandas as pd
import json
import random
import math
import numpy as np
import urllib3
import inspect
import folium
import pandas as pd
from IPython.display import display

def maximize_screen():
    from IPython.display import display, HTML
    display(HTML("<style>:root { --jp-notebook-max-width: 100% !important; }</style>"))

def timestamp():
    today=str(datetime.date.today()).split('-')
    year=today[0]
    month=today[1]
    day=today[2]
    a=str(datetime.datetime.now())
    time_split=re.split(':| ',a)
    hour=time_split[1]
    timestamp=year+month+day+hour
    return timestamp

maximize_screen()
timestamp=timestamp()
collection_date0=datetime.datetime.now().strftime("%m/%d/%Y")
collection_date_dt=pd.to_datetime(collection_date0)
collection_date=collection_date0


start_time=datetime.datetime.now()
event_time=start_time



# print('**Starting *** ',process_name,' *** , Time:',start_time,' **\n')
DirMain="C:\\Users\\jrozm\\OneDrive\\Benjamin\\"

fileInput="JewishCemeteryCoordinates"
FileInputCities="Jewish_Cemetery_Cities_Coordinates_X"
fileOutput="JewishCemetery_Site_Map"

# if append==1:fileIn='Jewish_Counties_Coordinates_X'
fileJewishIn=DirMain+"OutputData\\Cemeteries\\"+fileInput+".csv"
fileCitiesIn=DirMain+"Analysis\\Maps\\"+FileInputCities+".csv"

fileOut0=DirMain+"Analysis\\Maps\\"+fileOutput+".csv"
fileOut=DirMain+"Analysis\\Maps\\"+fileOutput+"_"+timestamp+".csv"
fileOutInt=DirMain+"Analysis\\Maps\\"+fileOutput+"_Interim.csv"



Jewish_Cemetery_Coordinates=pd.read_csv(fileJewishIn)
Jewish_Cemetery_Cities_Coordinates=pd.read_csv(fileCitiesIn)


cemetery_id_list=Jewish_Cemetery_Coordinates["cemetery_id"]
cemetery_name_list=Jewish_Cemetery_Coordinates["cemetery_name"]
county_id_list=Jewish_Cemetery_Coordinates["county_id"]
county_list=Jewish_Cemetery_Coordinates["county"]

grave_count_list=Jewish_Cemetery_Coordinates["grave_count"]
cemetery_city_list=Jewish_Cemetery_Coordinates["city"]

cemetery_lat_list=Jewish_Cemetery_Coordinates["cemetery_lat"]
cemetery_lon_list=Jewish_Cemetery_Coordinates["cemetery_lon"]
coordinates_list=Jewish_Cemetery_Coordinates[['cemetery_lat','cemetery_lon']]

city_list=Jewish_Cemetery_Cities_Coordinates["city"]
state_list=Jewish_Cemetery_Cities_Coordinates["state"]
city_lat_list=Jewish_Cemetery_Cities_Coordinates["City_Lat"]
city_lon_list=Jewish_Cemetery_Cities_Coordinates["City_Lon"]

field_list=['select_city','select_state','cemetery_name','cemetery_id','cemetery_city','cemetery_state','grave_count','distance']
city_distance_df = pd.DataFrame([])
city_cemetery_count=0
total_city_count=len(Jewish_Cemetery_Cities_Coordinates)

for city_count, city_row in Jewish_Cemetery_Cities_Coordinates.iterrows():
    city_center = (city_row["City_Lat"], city_row["City_Lon"])
    cemetery_count=0
    for cemetery_index, cemetery_row in Jewish_Cemetery_Coordinates.iterrows():
        city_distance= int(distance(city_center, (cemetery_row['cemetery_lat'], cemetery_row['cemetery_lon'])).miles)
        if city_distance>200: continue
        cemetery_count+=1
        city_distance_line_df=pd.DataFrame([city_row['city'],city_row['state'],cemetery_row['cemetery_name'],cemetery_row['cemetery_id'],cemetery_row['city'],\
                                      cemetery_row['state'],cemetery_row['grave_count'],city_distance]).T
        city_distance_line_df.columns=field_list
        city_distance_df = city_distance_df._append(city_distance_line_df, ignore_index = True) 
        city_cemetery_count+=1
    city_distance_df.to_csv(fileOutInt,mode="w", index=False,header=True)
    print(f"    ---Selected_City:  {city_row['city']},{city_row['state']} • City Count: {city_count+1}/{total_city_count} • Cemetery Count: {cemetery_count} ")
city_distance_df.to_csv(fileOut,mode="w", index=False,header=True)
city_distance_df.to_csv(fileOut0,mode="w", index=False,header=True) 
print("\n")
print('Uploaded To: ',fileOut0)    
   




# Find all cities within 100 miles

# within_radius = df[df['distance'] <= radius_miles]

# print(within_radius)
```
